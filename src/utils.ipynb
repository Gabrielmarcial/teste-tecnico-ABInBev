# MarTech Data Platform — Setup

Este notebook configura a estrutura inicial do data lakehouse para o domínio de **marketing technology (MarTech)**.

## Objetivo

Criar o catálogo `martech` e seus schemas organizados em camadas (medallion architecture) para dar suporte ao pipeline de dados de marketing:

| Camada | Schema | Finalidade |
|--------|--------|------------|
| Landing | `martech.raw` | Ingestão bruta dos dados de origem (source systems, APIs, CRM, etc.) |
| Bronze | `martech.bronze` | Dados brutos padronizados e versionados no Delta Lake |
| Silver | `martech.silver` | Dados limpos, validados e modelados (qualidade e consistência) |
| Gold | `martech.gold` | Dados prontos para consumo analítico e negócio (métricas, dashboards, ML) |

## Pré-requisitos

- Permissão `CREATE CATALOG` no metastore do Unity Catalog
- A execução é idempotente: os comandos usam `IF NOT EXISTS` e podem ser reexecutados sem erro


```python
%sql
CREATE CATALOG IF NOT EXISTS martech;
```

```python
%sql

CREATE SCHEMA IF NOT EXISTS martech.raw;
CREATE SCHEMA IF NOT EXISTS martech.bronze;
CREATE SCHEMA IF NOT EXISTS martech.silver;
CREATE SCHEMA IF NOT EXISTS martech.gold;
```

```python
%sql
CREATE VOLUME IF NOT EXISTS
martech.raw.raw_files;
```
