# MarTech Data Platform — Setup

Este notebook inicializa a infraestrutura do **data lakehouse para o domínio de marketing technology (MarTech)** no Unity Catalog.

## O que este notebook faz

- Cria o catálogo `martech`
- Cria os schemas em camadas (medallion architecture):
  - `martech.raw` — dados brutos ingeridos das fontes
  - `martech.bronze` — dados padronizados e versionados no Delta Lake
  - `martech.silver` — dados limpos, validados e modelados
  - `martech.gold` — dados prontos para consumo analítico e de negócio
- Cria os volumes auxiliares:
  - `raw_files` — armazenamento dos arquivos brutos ingeridos
  - `autoloader_metadata` e `checkpoint` — suporte aos pipelines de ingestão (Auto Loader / streaming)

## Observações

- **Idempotente**: todos os comandos usam `IF NOT EXISTS` e podem ser reexecutados sem erro
- **Pré-requisito**: permissão `CREATE CATALOG` no metastore do Unity Catalog
- Deve ser executado **antes** dos pipelines de ingestão e transformação

```python
%sql
CREATE CATALOG IF NOT EXISTS martech;
```

```python
%sql

CREATE SCHEMA IF NOT EXISTS martech.raw;
CREATE SCHEMA IF NOT EXISTS martech.bronze;
CREATE SCHEMA IF NOT EXISTS martech.silver;
CREATE SCHEMA IF NOT EXISTS martech.gold;
```
